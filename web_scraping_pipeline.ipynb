```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from datetime import datetime

# URL de exemplo (site seguro para testes de scraping: Quotes to Scrape)
url = "http://quotes.toscrape.com/"

print(f"🔄 [Scraper] A ligar ao site de destino: {url}")
response = requests.get(url)

if response.status_code != 200:
    raise Exception(f"Falha ao aceder à página: {response.status_code}")

# A processar o HTML da página
soup = BeautifulSoup(response.text, 'html.parser')

# Extraindo citações e autores para gerar inteligência de mercado/conteúdo
quotes_data = []
for quote_box in soup.find_all('div', class_='quote'):
    text = quote_box.find('span', class_='text').get_text()
    author = quote_box.find('small', class_='author').get_text()
    tags = [tag.get_text() for tag in quote_box.find_all('a', class_='tag')]

    quotes_data.append({
        'quote': text,
        'author': author,
        'tags': ", ".join(tags),
        'extracted_at': datetime.utcnow().strftime('%Y-%m-%d %H:%M:%S')
    })

# Transformando em DataFrame do Pandas
df = pd.DataFrame(quotes_data)

print(f"✅ [Scraper] Sucesso! {len(df)} registos extraídos da web!")
print("\n=== AMOSTRA DOS DADOS EXTRAÍDOS ===")
print(df.head(3))

# Salvando a base tratada em CSV
df.to_csv("harvested_web_data.csv", index=False)
print("\n📁 Dataset guardado localmente como 'harvested_web_data.csv'")
```
